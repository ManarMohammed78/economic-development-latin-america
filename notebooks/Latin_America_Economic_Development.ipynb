# Compare population in 2014 vs 2024 for all countries, and calculate growth

```python
population_pivot = population_df[population_df['Year'].isin([2014, 2024])].pivot(index='Country Name', columns='Year', values='Value')
```

```python
population_pivot['Growth %'] = ((population_pivot[2024] - population_pivot[2014]) / population_pivot[2014]) * 100
```

```python
population_pivot
```

# Compare population growth vs GDP per capita growth for each country (2014-2024)

```python
gdp_pivot = gdp_df[gdp_df['Year'].isin([2014, 2024])].pivot(index='Country Name', columns='Year', values='Value')
```

```python
gdp_pivot['GDP_Growth %'] = ((gdp_pivot[2024] - gdp_pivot[2014]) / gdp_pivot[2014]) * 100
```

```python
comparison = population_pivot[['Growth %']].rename(columns={'Growth %': 'Population_Growth %'})
```

```python
comparison['GDP_Growth %'] = gdp_pivot['GDP_Growth %']
```

```python
comparison
```

#15. **No inverse relationship between population growth and economic performance:** Contrary to the common assumption that rapid population growth strains resources and slows economic growth, the data shows no such pattern. Peru had both the highest population growth (13.6%) AND the highest GDP per capita growth (27.9%) in the period, while Brazil had the lowest population growth (5.95%) yet the worst GDP per capita performance (-16.0%) — indicating that other factors (e.g., economic policy, external shocks) played a far greater role than demographic pressure alone.
